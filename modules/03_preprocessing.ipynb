# 03 — Preprocessing
Normalize reflectance values, flatten the 3D cube, and reduce dimensions with PCA.

10 principal components capture **98.31%** of the variance in the original 256-band data.

```python
# Normalize to [0, 1] — improves model training stability
normalized_data = hyperspectral_data / np.max(hyperspectral_data)
print(f"Max after normalization: {np.max(normalized_data):.4f}")
```

```python
# Flatten 3D cube (rows × cols × bands) → 2D matrix (pixels × bands)
flattened_data = normalized_data.reshape(-1, normalized_data.shape[-1])
print(f"Flattened shape: {flattened_data.shape}")
```

## PCA — Dimensionality Reduction

```python
pca = PCA(n_components=10)
reduced_data = pca.fit_transform(flattened_data)

explained_variance = np.sum(pca.explained_variance_ratio_)
print(f"Explained variance by 10 components: {explained_variance:.4f}")
print(f"Reduced shape: {reduced_data.shape}")
```

```python
# Variance per component — shows how much each PC contributes
plt.figure(figsize=(10, 5))
plt.bar(range(1, 11), pca.explained_variance_ratio_ * 100, color='steelblue')
plt.xlabel('Principal Component')
plt.ylabel('Explained Variance (%)')
plt.title('PCA Explained Variance per Component')
plt.xticks(range(1, 11))
plt.grid(True, axis='y', alpha=0.3)
save_fig("pca_explained_variance")
plt.show()
```
