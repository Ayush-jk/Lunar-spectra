Visualize the hyperspectral strip at multiple bands and overlay geographical coordinates.

## Reflectance Variation at a Pixel
Enter any pixel coordinate `(x, y)` where x ∈ [0, 249] and y ∈ [0, 9339].

```python
def plot_reflectance_variation(x, y):
    if 0 <= x < hyperspectral_data.shape[1] and 0 <= y < hyperspectral_data.shape[0]:
        reflectance_values = hyperspectral_data[y, x, :]
        plt.figure(figsize=(10, 6))
        plt.plot(range(hyperspectral_data.shape[2]), reflectance_values.squeeze())
        plt.xlabel('Wavelength Band')
        plt.ylabel('Reflectance')
        plt.title(f'Reflectance Variation at ({x}, {y})')
        save_fig("reflectance_variation_vs_wavelength")
        plt.show()
    else:
        print("Invalid coordinates")
```

```python
x = int(input("Enter the x coordinate: "))
y = int(input("Enter the y coordinate: "))
plot_reflectance_variation(x, y)
```

## Grayscale Strip at 6 Bands
View the 9340×250 orbital strip at bands 10, 50, 100, 150, 200, 256.

```python
fig, axes = plt.subplots(1, 6, figsize=(15, 5))
for ax, band in zip(axes, [10, 50, 100, 150, 200, 255]):
    ax.imshow(hyperspectral_data[:, :, band], cmap='gray')
    ax.set_title(f'Band {band}')
save_fig("grayscale_image_6_bands")
plt.show()
```

## Region of Interest with Geographical Coordinates

```python
coord_df = pd.read_csv(geometry_file)
coord_df.head()
```

```python
roi_start_x, roi_end_x = 0, 250
roi_start_y, roi_end_y = 0, 250

def plot_roi_hyperspectral_with_coords(band_index):
    plt.figure(figsize=(20, 12))
    plt.imshow(hyperspectral_data[roi_start_y:roi_end_y, roi_start_x:roi_end_x, band_index], cmap='gray')
    for _, row in coord_df.iterrows():
        pixel, scan = row['Pixel'], row['Scan']
        if roi_start_x <= pixel < roi_end_x and roi_start_y <= scan < roi_end_y:
            plt.plot(pixel - roi_start_x, scan - roi_start_y, 'ro')
            plt.text(pixel - roi_start_x, scan - roi_start_y,
                     f"({row['Longitude']:.2f}, {row['Latitude']:.2f})", color='blue', fontsize=8)
    plt.title(f'Hyperspectral Data with Geographical Coordinates (Band {band_index}) - ROI')
    plt.xlabel('Pixel')
    plt.ylabel('Scan')
    save_fig("image_geographical_coordinates")
    plt.show()

plot_roi_hyperspectral_with_coords(int(input("Enter the band: ")))
```
