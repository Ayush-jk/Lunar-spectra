Cluster pixels into 5 spectral classes using K-Means on the PCA-reduced data.
These cluster assignments are used as labels for CNN training in the next notebook.

```python
kmean = KMeans(n_clusters=5, random_state=42)
labels = kmean.fit_predict(reduced_data)
print(f"Cluster counts: {dict(zip(*np.unique(labels, return_counts=True)))}")
```

## Manual K-Means Implementation (Reference)
The functions below implement K-Means from scratch for educational reference.
The sklearn implementation above is used for all downstream tasks.

```python
def nearest_centroid(X, centroid):
    k = centroid.shape[0]
    c = np.zeros(X.shape[0], dtype=int)
    for i in range(X.shape[0]):
        distances = [np.linalg.norm(X[i] - centroid[j]) for j in range(k)]
        c[i] = np.argmin(distances)
    return c

def compute_centroid(X, c, k):
    m, n = X.shape
    centroids = np.zeros((k, n))
    for i in range(k):
        centroids[i] = np.mean(X[c == i], axis=0)
    return centroids

def initialize_centroid(X, k):
    random_index = np.random.permutation(X.shape[0])
    return X[random_index[:k]]

print("Manual K-Means functions defined (reference implementation)")
```
